LSTM MODEL


```python

# Step 1: Import libraries
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from sklearn.model_selection import train_test_split

# Step 2: Prepare the reviews
reviews = [
    "The product quality is excellent",
    "I am very happy with the service",
    "The delivery was very slow",
    "The product is damaged",
    "Excellent customer support",
    "Very disappointed with the purchase"
]

# Positive = 1, Negative = 0
labels = np.array([1, 1, 0, 0, 1, 0])

# Step 3: Convert words into numerical sequences
tokenizer = Tokenizer()
tokenizer.fit_on_texts(reviews)

sequences = tokenizer.texts_to_sequences(reviews)

# Make all sequences the same length
max_length = 8
X = pad_sequences(sequences, maxlen=max_length, padding="post")

print("Numerical sequences:")
print(X)

# Step 4: Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, labels, test_size=0.33, random_state=42, stratify=labels
)

# Step 5: Build the LSTM model
model = Sequential([
    Embedding(
        input_dim=len(tokenizer.word_index) + 1,
        output_dim=16,
        input_length=max_length
    ),
    LSTM(16),
    Dense(1, activation="sigmoid")
])

# Step 6: Compile the model
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# Step 7: Train the model
model.fit(
    X_train, y_train,
    epochs=30,
    batch_size=2,
    verbose=1
)

# Step 8: Evaluate the model
loss, accuracy = model.evaluate(X_test, y_test, verbose=0)

print("\nTest Loss:", round(loss, 4))
print("Test Accuracy:", round(accuracy, 4))

# Step 9: Predict sentiment for a new review
new_review = ["The product is very good and easy to use"]

new_sequence = tokenizer.texts_to_sequences(new_review)
new_padded = pad_sequences(
    new_sequence, maxlen=max_length, padding="post"
)

probability = model.predict(new_padded, verbose=0)[0][0]

if probability >= 0.5:
    sentiment = "Positive"
else:
    sentiment = "Negative"

print("\nNew Review:", new_review[0])
print("Positive Probability:", round(float(probability), 4))
print("Predicted Sentiment:", sentiment)
```
