```python
corpus = [
    "<p>Data Science is AMAZING!!! Visit https://datascience.com</p>",
    "Python 3.12 is released in 2025.",
    "NLP is used in AI.",
    "Contact us at info@gmail.com"
]
```

```python
import re
from bs4 import BeautifulSoup
```

```python
lowercase_corpus = []

for text in corpus:
    lowercase_corpus.append(text.lower())
print(lowercase_corpus)
```

```python
html_removed_corpus = []

for text in lowercase_corpus:
    clean_text = re.sub(r"<.*?>", "", text)
    html_removed_corpus.append(clean_text)
print(html_removed_corpus)
```

```python
#5Remove URLs.
url_removed_corpus = []

for text in html_removed_corpus:
    clean = re.sub(r"http\S+|www\S+", "", text)
    url_removed_corpus.append(clean)
print(url_removed_corpus)
```

```python
#Remove Email Addresses
email_removed_corpus = []

for text in url_removed_corpus:
    clean = re.sub(r"\S+@\S+", "", text)
    email_removed_corpus.append(clean)
print(email_removed_corpus)
```

```python
#Remove digits
digit_removed_corpus = []

for text in email_removed_corpus:
    clean = re.sub(r"\d+", "", text)
    digit_removed_corpus.append(clean)
print(digit_removed_corpus)
```

```python
import re

text = "Hello!!!"

clean_text = re.sub(r"[^\w\s]", "", text)

print(clean_text)
```

```python
#Remove extra white spaces
whitespace_removed_corpus = []
for text in digit_removed_corpus:
    clean = re.sub(r"\s+", " ", text).strip()
    whitespace_removed_corpus.append(clean)
print(whitespace_removed_corpus)
```

```python
#Store in cleaned_corpus
cleaned_corpus = whitespace_removed_corpus
print(cleaned_corpus)
```

```python
#Display the Original Corpus and the Cleaned Corpus
print("Original Corpus:\n")

for text in corpus:
    print(text)

print("\nCleaned Corpus:\n")

for text in cleaned_corpus:
    print(text)
```

```python
#Compare the Text Before and After Cleaning
print("Comparison of Original and Cleaned Text:\n")

for original, cleaned in zip(corpus, cleaned_corpus):
    print("Original :", original)
    print("Cleaned  :", cleaned)
    print("-" * 50)
```
