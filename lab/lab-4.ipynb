# Activity 1: Vocabulary Creation

## Aim
To create and display the vocabulary from a given text corpus.

## Objectives
1. Import the required Python libraries.
2. Create the input corpus.
3. Generate the vocabulary.
4. Display the vocabulary.

```python
# Activity 1: Vocabulary Creation

# Step 1: Import required library
import re

# Step 2: Create the input corpus
documents = [
    "I love machine learning",
    "I love Python programming",
    "Machine learning is useful"
]

# Step 3: Generate the vocabulary
vocabulary = set()

for document in documents:
    words = document.lower().split()
    vocabulary.update(words)

vocabulary = sorted(vocabulary)

# Step 4: Display the vocabulary
print("Input Corpus:")
for i, document in enumerate(documents, 1):
    print(f"Document {i}: {document}")

print("\nVocabulary:")
print(vocabulary)

print("\nTotal Vocabulary Size:", len(vocabulary))

print("\nVocabulary with Index:")
for i, word in enumerate(vocabulary, 1):
    print(i, word)
```

# Activity 2: TF, DF, IDF and TF-IDF Matrix

## Aim
To calculate Term Frequency (TF), Document Frequency (DF), Inverse Document Frequency (IDF), and generate the TF-IDF matrix.

## Objectives
1. Calculate the total number of words in each document.
2. Compute Term Frequency (TF).
3. Display the TF matrix.
4. Calculate Document Frequency (DF).
5. Compute Inverse Document Frequency (IDF).
6. Display DF and IDF values and the TF-IDF matrix.

```python
# Activity 2: TF, DF, IDF and TF-IDF Matrix

# Step 1: Import required libraries
import math
import pandas as pd

# Step 2: Create the input corpus
documents = [
    "I love machine learning",
    "I love Python programming",
    "Machine learning is useful"
]

# Convert documents to lowercase and tokenize
tokenized_documents = [
    document.lower().split()
    for document in documents
]

# Step 3: Generate vocabulary
vocabulary = sorted(
    set(word for document in tokenized_documents for word in document)
)

print("Vocabulary:")
print(vocabulary)

# Step 4: Calculate total number of words in each document
total_words = [
    len(document)
    for document in tokenized_documents
]

print("\nTotal Number of Words:")
for i, count in enumerate(total_words, 1):
    print(f"Document {i}: {count}")

# Step 5: Calculate Term Frequency (TF)
tf_matrix = []

for document in tokenized_documents:
    total = len(document)
    tf_values = []

    for word in vocabulary:
        count = document.count(word)
        tf = count / total
        tf_values.append(tf)

    tf_matrix.append(tf_values)

tf_df = pd.DataFrame(
    tf_matrix,
    columns=vocabulary,
    index=["Document 1", "Document 2", "Document 3"]
)

print("\nTF Matrix:")
display(tf_df.round(4))

# Step 6: Calculate Document Frequency (DF)
df_values = []

for word in vocabulary:
    df = sum(
        word in document
        for document in tokenized_documents
    )
    df_values.append(df)

# Step 7: Calculate Inverse Document Frequency (IDF)
number_of_documents = len(tokenized_documents)

idf_values = []

for df in df_values:
    idf = math.log(number_of_documents / df)
    idf_values.append(idf)

df_idf_table = pd.DataFrame({
    "Term": vocabulary,
    "DF": df_values,
    "IDF": idf_values
})

print("\nDF and IDF Values:")
display(df_idf_table.round(4))

# Step 8: Calculate TF-IDF
tfidf_matrix = []

for i in range(number_of_documents):
    row = []

    for j in range(len(vocabulary)):
        tfidf = tf_matrix[i][j] * idf_values[j]
        row.append(tfidf)

    tfidf_matrix.append(row)

tfidf_df = pd.DataFrame(
    tfidf_matrix,
    columns=vocabulary,
    index=["Document 1", "Document 2", "Document 3"]
)

print("\nTF-IDF Matrix:")
display(tfidf_df.round(4))

# Conclusion
print("\nTF-IDF calculation completed successfully!")
```
