```python
# Step 1: Import required libraries
from gensim.models import Word2Vec
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# Step 2: Create a small text corpus
corpus = [
    "I love machine learning",
    "I enjoy artificial intelligence",
    "Cats drink milk",
    "Dogs drink milk"
]

# Step 3: Tokenize the corpus
tokenized_corpus = [sentence.lower().split() for sentence in corpus]

# Step 4: Train the Word2Vec model
model = Word2Vec(
    sentences=tokenized_corpus,
    vector_size=100,
    window=2,
    min_count=1,
    sg=1
)

# Step 5 & 6: Convert each sentence into a sentence embedding
def sentence_embedding(sentence):
    words = sentence.lower().split()
    vectors = [model.wv[word] for word in words]
    return np.mean(vectors, axis=0)

sentence1 = "I love machine learning"
sentence2 = "I enjoy artificial intelligence"

vector1 = sentence_embedding(sentence1)
vector2 = sentence_embedding(sentence2)

# Step 7: Compute cosine similarity
similarity = cosine_similarity([vector1], [vector2])[0][0]

# Step 8: Display the similarity score
print("Sentence 1 :", sentence1)
print("Sentence 2 :", sentence2)
print("Cosine Similarity :", round(similarity, 2))
```
