```python

# Step 1: Import required libraries
from gensim.models import Word2Vec
from nltk.tokenize import wordpunct_tokenize

# Step 2: Create the sample corpus
sentences = [
    "Machine learning is a branch of artificial intelligence.",
    "Artificial intelligence helps computers learn from data.",
    "Deep learning is a subset of machine learning.",
    "Natural language processing uses machine learning."
]

# Step 3: Tokenize the sentences
tokenized_sentences = []

for sentence in sentences:
    words = wordpunct_tokenize(sentence.lower())
    # Remove punctuation
    words = [word for word in words if word.isalnum()]
    tokenized_sentences.append(words)

print("Tokenized Corpus:")
for sentence in tokenized_sentences:
    print(sentence)

# Step 4: Train Word2Vec using CBOW (sg=0)
model = Word2Vec(
    sentences=tokenized_sentences,
    vector_size=50,
    window=2,
    min_count=1,
    sg=0,
    epochs=100,
    seed=42
)

print("\nModel trained successfully!")

# Step 5: Display the vocabulary
print("\nVocabulary:")
print(list(model.wv.index_to_key))

# Step 6: Display word embeddings
print("\nVector for 'machine':")
print(model.wv["machine"])

print("\nVector for 'learning':")
print(model.wv["learning"])

print("\nVector dimension:", model.wv.vector_size)

# Step 7: Find the top 5 similar words
print("\nTop 5 words similar to 'learning':")
similar_words = model.wv.most_similar("learning", topn=5)

for word, score in similar_words:
    print(word, ":", round(score, 4))

# Step 8: Calculate semantic similarity
print("\nSimilarity between Machine and Learning:")
print(round(model.wv.similarity("machine", "learning"), 4))

print("\nSimilarity between Artificial and Intelligence:")
print(round(model.wv.similarity("artificial", "intelligence"), 4))
```
