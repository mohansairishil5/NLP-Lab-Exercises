```python
# Activity 1: Spam Detection using Naïve Bayes

# Step 1: Import required libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, confusion_matrix

# Step 2: Create the spam email dataset
emails = [
    "Congratulations you won a free prize",
    "Claim your free gift now",
    "Win a cash reward today",
    "You have won a lottery",
    "Get free money immediately",
    "Exclusive offer click now",
    "Free entry in contest",
    "Win a free mobile phone",
    "Meeting is scheduled for tomorrow",
    "Please submit the project report",
    "Your assignment is due today",
    "Let's discuss the project",
    "The meeting starts at 10 AM",
    "Please attend the class tomorrow",
    "Your report has been received",
    "Can we meet today to discuss work"
]

labels = [
    "spam", "spam", "spam", "spam",
    "spam", "spam", "spam", "spam",
    "ham", "ham", "ham", "ham",
    "ham", "ham", "ham", "ham"
]

# Step 3: Convert emails into TF-IDF feature vectors
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(emails)
y = labels

# Step 4: Split dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

# Step 5: Train the Naïve Bayes classifier
model = MultinomialNB()
model.fit(X_train, y_train)

# Step 6: Predict the class of test emails
y_pred = model.predict(X_test)

# Step 7: Display accuracy
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)

# Step 8: Display confusion matrix
cm = confusion_matrix(y_test, y_pred)

print("\nConfusion Matrix:")
print(cm)

# Step 9: Predict the class of new email messages
new_emails = [
    "Congratulations you won free money",
    "Please submit your assignment tomorrow",
    "Claim your free prize now"
]

new_X = vectorizer.transform(new_emails)
new_predictions = model.predict(new_X)

print("\nNew Email Predictions:")
for email, prediction in zip(new_emails, new_predictions):
    print("Email:", email)
    print("Prediction:", prediction)
    print()
```

```python
# Activity 2: Sentiment Analysis using SVM

# Step 1: Import required libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix

# Step 2: Create the customer review dataset
reviews = [
    "The product is excellent",
    "I really love this product",
    "Amazing quality and great service",
    "The product is very good",
    "I am happy with my purchase",
    "Excellent experience",
    "The service was wonderful",
    "I highly recommend this product",
    "The product is terrible",
    "I hate this product",
    "Very poor quality",
    "The service was bad",
    "I am disappointed with the product",
    "Worst experience ever",
    "The product is useless",
    "I would not recommend this product"
]

sentiments = [
    "positive", "positive", "positive", "positive",
    "positive", "positive", "positive", "positive",
    "negative", "negative", "negative", "negative",
    "negative", "negative", "negative", "negative"
]

# Step 3: Convert reviews into TF-IDF feature vectors
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(reviews)
y = sentiments

# Step 4: Split dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

# Step 5: Train the SVM classifier
model = SVC(kernel="linear")
model.fit(X_train, y_train)

# Step 6: Predict the sentiment of test reviews
y_pred = model.predict(X_test)

# Step 7: Display accuracy
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)

# Step 8: Display confusion matrix
cm = confusion_matrix(
    y_test,
    y_pred,
    labels=["positive", "negative"]
)

print("\nConfusion Matrix:")
print(cm)

# Step 9: Predict the sentiment of new reviews
new_reviews = [
    "The product is amazing",
    "I am very disappointed",
    "Excellent quality and service"
]

new_X = vectorizer.transform(new_reviews)
new_predictions = model.predict(new_X)

print("\nNew Review Predictions:")
for review, prediction in zip(new_reviews, new_predictions):
    print("Review:", review)
    print("Sentiment:", prediction)
    print()
```
