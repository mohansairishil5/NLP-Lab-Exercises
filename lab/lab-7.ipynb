```python

import numpy as np
import gensim.downloader as api
from sklearn.metrics.pairwise import cosine_similarity

# Step 1: Load the pre-trained GloVe model
print("Loading GloVe model...")
glove_model = api.load("glove-wiki-gigaword-50")

# Step 2: Define two input sentences
sentence1 = "I love machine learning"
sentence2 = "I enjoy artificial intelligence"

# Step 3: Convert a sentence into an embedding
def sentence_embedding(sentence):
    words = sentence.lower().split()

    # Get GloVe vectors for words present in the model
    word_vectors = [
        glove_model[word]
        for word in words
        if word in glove_model.key_to_index
    ]

    # Average the word vectors
    if len(word_vectors) == 0:
        return np.zeros(glove_model.vector_size)

    return np.mean(word_vectors, axis=0)

# Step 4: Generate sentence embeddings
vector1 = sentence_embedding(sentence1)
vector2 = sentence_embedding(sentence2)

# Step 5: Calculate cosine similarity
similarity = cosine_similarity(
    vector1.reshape(1, -1),
    vector2.reshape(1, -1)
)[0][0]

# Step 6: Display the result
print("\nSentence 1:", sentence1)
print("Sentence 2:", sentence2)
print("Cosine Similarity:", round(float(similarity), 2))
```
